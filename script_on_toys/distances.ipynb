# Centroid MASs

```python
d=3
k1,k2=4,5
nb_MAS1,nb_MAS2=10,11 #nb streamline
nb_MAP1,nb_MAP2=20,21 #nb points on streamline
m1,wI1,I1,w1,S1=random_MASs(nb_MAS1,nb_MAP1,k1,d)
m2,wI2,I2,w2,S2=random_MASs(nb_MAS2,nb_MAP2,k2,d)
print(m1.shape,wI1.shape,I1.shape,w1.shape,S1.shape)
print(m2.shape,wI2.shape,I2.shape,w2.shape,S2.shape)
```

```python
# Distance between two covariates
print(distance_aniso(S1[0,0,0],S2[0,0,0]))

#matrix of distances between two covariates
print(distances_aniso(S1[0,0],S2[0,0]).shape)
print(distances_aniso(S1[0,0],S2[0,0])[0,0])

#matrices of distances between two covariates along a streamline
print(distances_aniso_along(S1[0],S2[0,:20]).shape)
print(distances_aniso_along(S1[0],S2[0,:20])[0,0,0])
```

```python
#matrix of distances between two covariates and isotropic compartment
print(distances_iso_aniso(I1[0,0],I2[0,0],S1[0,0],S2[0,0]))

#matrix of distances between two covariates and isotropic compartment along a streamline
print(distances_iso_aniso_along(I1[0],I2[0,:20],S1[0],S2[0,:20])[0])
print(distances_iso_aniso_along(I1[0],I2[0,:20],S1[0],S2[0,:20]).shape)
```

```python
#matrix of distances between two spatials, covariates and isotropic compartment
print(distances_spatial_iso_aniso(m1[0,0],m2[0,0],I1[0,0],I2[0,0],S1[0,0],S2[0,0]))
print(distances_spatial_iso_aniso(m1[0,0],m2[0,0],I1[0,0],I2[0,0],S1[0,0],S2[0,0]).shape)

#matrix of distances between two spatials, covariates and isotropic compartment along a streamline
print(distances_spatial_iso_aniso_along(m1[0],m2[0,:20],I1[0],I2[0,:20],S1[0],S2[0,:20])[0])
print(distances_spatial_iso_aniso_along(m1[0],m2[0,:20],I1[0],I2[0,:20],S1[0],S2[0,:20]).shape)
```

```python
# distance between 2 MAPs 
print(distance_MAP(m1[0,0],m2[0,0],wI1[0,0],wI2[0,0],I1[0,0],I2[0,0],w1[0,0],w2[0,0],S1[0,0],S2[0,0],alpha=.5))

# Matrix of distance between 2 MAPs 
print(distances_MAP(m1[0],m2[0],wI1[0],wI2[0],I1[0],I2[0],w1[0],w2[0],S1[0],S2[0],alpha=.5).shape)
print(distances_MAP(m1[0],m2[0],wI1[0],wI2[0],I1[0],I2[0],w1[0],w2[0],S1[0],S2[0],alpha=.5)[0])
```

```python
#Distance between 2 MAS (with aligned indices)
print(distance_MAS_idx(m1[0],m2[0,:20],wI1[0],wI2[0,:20],I1[0],I2[0,:20],w1[0],w2[0,:20],S1[0],S2[0,:20],alpha=.5))

#Distance between 2 MAS
print(distance_MAS(m1[0],m2[0],wI1[0],wI2[0],I1[0],I2[0],w1[0],w2[0],S1[0],S2[0],alpha=.5))

#Matrix of distance between MAS
print(distances_MAS_idx(m1,m2[:,:20],wI1,wI2[:,:20],I1,I2[:,:20],w1,w2[:,:20],S1,S2[:,:20],alpha=0.5).shape)
```

```python
#Distance between 2 similar tracts
print(distance_MAT_idx(m1,m1,wI1,wI1,I1,I1,w1,w1,S1,S1,alpha=.5,reflexion=False,lam1=0,lam2=0))


#Distance between 2 tracts
print(distance_MAT_idx(m1,m2[:10,:20],wI1,wI2[:10,:20],I1,I2[:10,:20],w1,w2[:10,:20],S1,S2[:10,:20],alpha=.5,reflexion=False,lam1=0,lam2=0))
```
